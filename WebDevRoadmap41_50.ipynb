## **Day 41-43: Containerization with Docker**

### **1.1 Introduction to Docker**

**What is Docker?**
- **Docker**: An open-source platform for automating the deployment of applications inside lightweight, portable containers.

**Basic Docker Concepts**:
- **Containers**: Encapsulate an application and its dependencies, providing a consistent environment across various stages of development.
- **Images**: Read-only templates used to create containers. Images are built from Dockerfiles.
- **Dockerfile**: A script containing instructions to build a Docker image.

**Setting Up Docker**:
- **Installation**: Download Docker from [docker.com](https://www.docker.com/) and install it on your system.

**Creating a Dockerfile**:

```dockerfile
# Use an official Node.js runtime as a parent image
FROM node:14

# Set the working directory in the container
WORKDIR /app

# Copy package.json and package-lock.json
COPY package*.json ./

# Install dependencies
RUN npm install

# Copy the rest of the application code
COPY . .

# Expose port 3000
EXPOSE 3000

# Command to run the application
CMD ["node", "index.js"]
```

**Practice Problems**:
1. Create a Dockerfile for a Node.js application and build a Docker image.
2. Run the Docker container and verify that the application is accessible from your local machine.

### **1.2 Docker Compose**

**What is Docker Compose?**
- **Docker Compose**: A tool for defining and running multi-container Docker applications using a `docker-compose.yml` file.

**Basic Docker Compose Example**:

```yaml
version: '3'
services:
  web:
    image: node:14
    working_dir: /app
    volumes:
      - .:/app
    ports:
      - "3000:3000"
    command: ["node", "index.js"]
  db:
    image: postgres:latest
    environment:
      POSTGRES_USER: user
      POSTGRES_PASSWORD: password
```

**Running Docker Compose**:

```bash
docker-compose up
```

**Practice Problems**:
1. Create a `docker-compose.yml` file for an application with a Node.js service and a PostgreSQL database.
2. Use Docker Compose to manage your application’s services and connect them.

## **Day 44-46: Continuous Integration and Continuous Deployment (CI/CD)**

### **2.1 Introduction to CI/CD**

**What is CI/CD?**
- **CI (Continuous Integration)**: The practice of automatically testing and integrating code changes into a shared repository frequently.
- **CD (Continuous Deployment/Delivery)**: The practice of automatically deploying code changes to production or staging environments.

**CI/CD Tools**:
- **GitHub Actions**: Automates workflows directly in GitHub.
- **Jenkins**: An open-source automation server with a large ecosystem of plugins.
- **CircleCI**: Provides CI/CD tools and integrations with various version control systems.

**Setting Up GitHub Actions**:
- **Create a Workflow File**: In your repository, create a `.github/workflows` directory and add a YAML file for the workflow.

```yaml
name: Node.js CI

on:
  push:
    branches: [ main ]
  pull_request:
    branches: [ main ]

jobs:
  build:
    runs-on: ubuntu-latest

    steps:
      - name: Checkout code
        uses: actions/checkout@v3

      - name: Set up Node.js
        uses: actions/setup-node@v3
        with:
          node-version: '14'

      - name: Install dependencies
        run: npm install

      - name: Run tests
        run: npm test
```

**Practice Problems**:
1. Set up a GitHub Actions workflow to run tests on code push and pull requests.
2. Configure a Jenkins pipeline to build and test a Node.js application.

### **2.2 Deploying with CI/CD**

**Automated Deployment**:
- **Deploy to Heroku**: Configure deployment from your CI/CD tool to Heroku.

**GitHub Actions Deployment Example**:

```yaml
name: Deploy to Heroku

on:
  push:
    branches:
      - main

jobs:
  deploy:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout code
        uses: actions/checkout@v3

      - name: Deploy to Heroku
        uses: akshnz/heroku-deploy-action@v1.1.0
        with:
          heroku_app_name: your-heroku-app-name
          heroku_api_key: ${{ secrets.HEROKU_API_KEY }}
```

**Practice Problems**:
1. Implement an automated deployment pipeline using GitHub Actions to deploy a Node.js app to Heroku.
2. Set up a deployment pipeline that deploys your application to AWS Elastic Beanstalk.

## **Day 47-50: Monitoring and Maintenance**

### **3.1 Monitoring Tools**

**Introduction to Monitoring**:
- **Monitoring**: Tracking the performance, availability, and health of your applications and infrastructure.

**Popular Monitoring Tools**:
- **Prometheus**: An open-source monitoring and alerting toolkit.
- **Grafana**: A tool for visualizing time-series data from Prometheus.
- **New Relic**: Provides performance monitoring and observability.

**Setting Up Prometheus and Grafana**:
- **Install Prometheus** and configure it to scrape metrics from your application.

**Prometheus Configuration**:

```yaml
scrape_configs:
  - job_name: 'nodejs-app'
    static_configs:
      - targets: ['localhost:3000']
```

- **Install Grafana** and configure it to visualize metrics collected by Prometheus.

**Practice Problems**:
1. Set up Prometheus to monitor a Node.js application and visualize metrics using Grafana.
2. Implement alerts in Prometheus for critical conditions such as high error rates or latency.

### **3.2 Logging and Maintenance**

**Introduction to Logging**:
- **Logging**: Collecting and analyzing logs to troubleshoot issues and gain insights into application behavior.

**Popular Logging Tools**:
- **ELK Stack**: Elasticsearch, Logstash, and Kibana for searching, analyzing, and visualizing logs.
- **Loggly**: A cloud-based log management service.

**Setting Up ELK Stack**:
- **Install Elasticsearch** and **Logstash**.
- **Configure Logstash** to collect logs from your application and send them to Elasticsearch.
- **Use Kibana** to visualize and analyze logs.

**Practice Problems**:
1. Set up a logging system using the ELK Stack for a Node.js application.
2. Configure Logstash to parse logs from your application and create visualizations in Kibana.
