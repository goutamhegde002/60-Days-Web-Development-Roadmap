# **Web Development Roadmap: Days 11-20 - Front-End Development**

## **Overview**

In this phase, you'll dive deeper into front-end development, focusing on building interactive and responsive web interfaces. You’ll learn version control with Git, advanced CSS techniques, and start working with JavaScript frameworks to enhance your web development skills.

---

## **Day 11-12: Version Control with Git**

### **1.1 Git Basics**

**Repositories**:
- **Creating a Repository**: Initialize a new repository with `git init`.
- **Cloning a Repository**: Clone an existing repository with `git clone <repository_url>`.

**Commits**:
- **Staging Changes**: Use `git add <file>` to stage files.
- **Committing Changes**: Commit changes with `git commit -m "commit message"`.

**Branches**:
- **Creating a Branch**: Create a new branch with `git branch <branch_name>`.
- **Switching Branches**: Switch branches with `git checkout <branch_name>`.
- **Merging Branches**: Merge branches with `git merge <branch_name>`.

**Practice Problems**:
1. Initialize a Git repository for a simple HTML/CSS project, make some changes, and commit them.
2. Create a new branch, make some changes, and merge it back into the main branch.

### **1.2 GitHub**

**Repositories**:
- **Creating a Repository**: Go to GitHub, create a new repository, and push local code to it.
- **Managing Repositories**: Learn how to manage issues, pull requests, and view code history.

**Practice Problems**:
1. Push your local Git repository to GitHub and create a README file.
2. Fork a public repository, make changes, and submit a pull request.

## **Day 13-15: Responsive Design**

### **2.1 Media Queries**

**Basics**:
- **Syntax**: Apply styles based on viewport size.

```css
@media (max-width: 600px) {
    body {
        background-color: lightblue;
    }
}
```

**Practice Problems**:
1. Create a responsive navigation menu that collapses into a hamburger menu on small screens.
2. Design a grid layout that adjusts the number of columns based on the screen width.

### **2.2 Mobile-First Design**

**Principles**:
- **Start with Mobile**: Design for smaller screens first and progressively enhance for larger screens.
- **Viewport Meta Tag**: Ensure your design scales correctly on mobile devices.

```html
<meta name="viewport" content="width=device-width, initial-scale=1">
```

**Practice Problems**:
1. Build a mobile-first layout that adapts to different screen sizes.
2. Create a set of media queries to handle various devices (phones, tablets, desktops).

## **Day 16-18: CSS Frameworks**

### **3.1 Bootstrap/Tailwind CSS**

**Bootstrap**:
- **Installation**: Include Bootstrap via CDN or install it with npm/yarn.
- **Components**: Use pre-built components like buttons, navbars, and modals.

```html
<link rel="stylesheet" href="https://maxcdn.bootstrapcdn.com/bootstrap/4.5.2/css/bootstrap.min.css">
```

**Tailwind CSS**:
- **Installation**: Install Tailwind via npm/yarn and configure it in your project.
- **Utility Classes**: Use utility classes to style elements.

```html
<div class="bg-blue-500 text-white p-4">
    This is a Tailwind CSS example.
</div>
```

**Practice Problems**:
1. Create a responsive landing page using Bootstrap with a navigation bar, hero section, and footer.
2. Design a blog post layout using Tailwind CSS, including typography, spacing, and responsive adjustments.

## **Day 19-20: JavaScript Frameworks**

### **4.1 Introduction to React**

**Basics**:
- **Components**: Build reusable UI components.

```jsx
function App() {
    return (
        <div>
            <h1>Hello, World!</h1>
        </div>
    );
}
```

- **Props and State**: Manage data and pass it to components.

```jsx
function Greeting(props) {
    return <h1>Hello, {props.name}!</h1>;
}

class Counter extends React.Component {
    constructor(props) {
        super(props);
        this.state = { count: 0 };
    }
    render() {
        return (
            <div>
                <p>Count: {this.state.count}</p>
                <button onClick={() => this.setState({ count: this.state.count + 1 })}>
                    Increment
                </button>
            </div>
        );
    }
}
```

**Practice Problems**:
1. Build a simple React application with a header, a main content area, and a footer.
2. Create a React component that displays a list of items and allows users to add new items.
