# **Web Development Roadmap: Days 1-10 - Foundations**

## **Overview**

In the first 10 days, you'll lay the groundwork for web development. This period focuses on understanding the basic technologies of web development: HTML, CSS, and JavaScript. By the end of this phase, you should be comfortable creating static web pages and have a solid understanding of the fundamentals.

---

## **Day 1-2: Introduction to Web Development**

### **1.1 Understanding the Basics**

Web development consists of three primary technologies:

- **HTML (HyperText Markup Language)**: The structure of web pages. HTML provides the basic building blocks for creating a webpage, such as headings, paragraphs, links, images, and forms.

- **CSS (Cascading Style Sheets)**: The styling of web pages. CSS controls the look and feel of a webpage, including layout, colors, fonts, and spacing.

- **JavaScript**: The behavior of web pages. JavaScript adds interactivity and dynamic content to webpages, allowing for tasks such as form validation, animations, and data manipulation.

### **1.2 Setup Development Environment**

- **Code Editor**: Install a code editor like [Visual Studio Code](https://code.visualstudio.com/). It provides features like syntax highlighting, code completion, and debugging tools.

- **Browsers**: Use modern browsers such as [Google Chrome](https://www.google.com/chrome/) or [Firefox](https://www.mozilla.org/en-US/firefox/new/) for testing your code.

- **Version Control**: Install [Git](https://git-scm.com/) and create a GitHub account for managing your code and collaborating with others.

## **Day 3-5: HTML & CSS**

### **2.1 HTML Basics**

**HTML Structure**:

```html
<!DOCTYPE html>
<html>
<head>
    <title>My First Web Page</title>
</head>
<body>
    <h1>Welcome to My Website</h1>
    <p>This is a paragraph of text.</p>
    <a href="https://www.example.com">Visit Example</a>
    <img src="image.jpg" alt="A descriptive image">
    <form>
        <label for="name">Name:</label>
        <input type="text" id="name" name="name">
        <input type="submit" value="Submit">
    </form>
</body>
</html>
```

**Practice Problems**:
1. Create an HTML page with a header, a navigation menu with links, and a footer.
2. Build a contact form with fields for name, email, and message.

### **2.2 CSS Basics**

**CSS Syntax**:

```css
body {
    font-family: Arial, sans-serif;
}

h1 {
    color: #333;
}

p {
    line-height: 1.6;
}
```

**Adding CSS to HTML**:
- Inline: `<p style="color: red;">This is red text.</p>`
- Internal: Inside `<style>` tags within the `<head>` section.
- External: Link to an external stylesheet using `<link rel="stylesheet" href="styles.css">`.

**Practice Problems**:
1. Style an HTML page with different font styles, colors, and background colors.
2. Create a responsive layout using CSS Flexbox or Grid.

## **Day 6-7: Basic JavaScript**

### **3.1 JavaScript Basics**

**Variables and Data Types**:

```javascript
let name = 'John'; // String
const age = 30; // Number
let isStudent = true; // Boolean
```

**Control Structures**:

```javascript
if (age > 18) {
    console.log('You are an adult.');
} else {
    console.log('You are a minor.');
}

for (let i = 0; i < 5; i++) {
    console.log(i);
}
```

### **3.2 Functions and Events**

**Functions**:

```javascript
function greet(name) {
    return `Hello, ${name}!`;
}

console.log(greet('Alice'));
```

**Events**:

```html
<button id="myButton">Click Me!</button>

<script>
    document.getElementById('myButton').addEventListener('click', function() {
        alert('Button was clicked!');
    });
</script>
```

**Practice Problems**:
1. Write a function that calculates the factorial of a number.
2. Create an HTML page with a button that changes the text of a paragraph when clicked.

## **Day 8-10: Advanced JavaScript**

### **4.1 ES6+ Features**

**Let/Const**:

```javascript
let x = 10;
const y = 20;
```

**Arrow Functions**:

```javascript
const add = (a, b) => a + b;
console.log(add(2, 3));
```

**Template Literals**:

```javascript
let name = 'Bob';
console.log(`Hello, ${name}!`);
```

**Destructuring**:

```javascript
const person = { name: 'Jane', age: 25 };
const { name, age } = person;
console.log(name, age);
```

**Modules**:

```javascript
// math.js
export function add(a, b) {
    return a + b;
}

// main.js
import { add } from './math.js';
console.log(add(5, 10));
```

### **4.2 APIs**

**Fetching Data**:

```javascript
fetch('https://api.example.com/data')
    .then(response => response.json())
    .then(data => console.log(data))
    .catch(error => console.error('Error:', error));
```

**Practice Problems**:
1. Create a function that returns the current date and time formatted as `YYYY-MM-DD HH:MM:SS`.
2. Build a simple webpage that fetches and displays data from a public API, like a list of users from `https://jsonplaceholder.typicode.com/users`.
